```python
import os
import pandas as pd
from pathlib import Path

DATADIR = Path("../../../data")
```

```python
# process Wang2025 raw download script
metadata = pd.read_csv(
    DATADIR / "raw/Wang2025/filereport_read_run_SRP348401.tsv", sep="\t"
)
metadata = metadata[metadata["scientific_name"] == "Mus musculus"]

search = "|".join(metadata.run_accession.tolist())

with open(
    DATADIR
    / "raw/Wang2025/ena-file-download-read_run-SRP348401-fastq_ftp-20260116-1907.sh",
    "r",
    encoding="utf-8",
) as file:
    content = file.read()
downloads = pd.Series(content.split("\n"))
downloads[downloads.str.contains(search)].to_csv(
    DATADIR / "raw/Wang2025/ena-download-edited.sh"
)
```

```python
# get data for Wang2025

# !wget -nc ftp://ftp.ncbi.nlm.nih.gov/geo/series/GSE189nnn/GSE189783/suppl/GSE189783_RAW.tar
# !tar -xvf GSE189783_RAW.tar

files = list((DATADIR / "cellranger" / "Wang2025_download").glob("*"))
files = pd.Series([str(file) for file in files]).sort_values()
for sample in files.str.extract(r"(GSM[0-9]{7}_.*\d)")[0].unique():
    os.makedirs(DATADIR / "cellranger" / "Wang2025_download" / sample, exist_ok=True)
    for file in files[files.str.contains(sample)]:
        newname = file.split("/")
        newname.insert(-1, sample)
        newname[-1] = newname[-1].split(sample + "_")[0]
        newname = "/".join(newname)
        os.rename(file, newname)
```

```python
for sample in (DATADIR / "cellranger" / "Wang2025_download").glob("*"):
    print(os.path.basename(sample))
    for file in os.listdir(sample):
        os.rename(
            sample / file, sample / file.split(os.path.basename(sample) + "_")[-1]
        )
```
